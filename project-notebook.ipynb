## 6. Design System (shadcn/ui)

### Status: Applied ✅

Scaffolded via:
```
npx shadcn@latest init --preset b3SkLI2EK --template next
```

This is now the project's official design system. All future interfaces and
components should be built with it, and any uploaded/pasted/generated designs
should be adapted to it rather than introducing new styling systems. This rule
is codified in `CLAUDE.md` at the repo root.

### What it set up
- `components.json` — style `base-sera`, base color `neutral`, CSS variables on,
  icon library `lucide`
- Tailwind CSS v4 installed (`postcss.config.mjs`, `@import "tailwindcss"` in
  `app/globals.css`)
- `lib/utils.ts` — `cn()` helper (clsx + tailwind-merge)
- `components/ui/button.tsx` — first installed component
- Component primitives come from `@base-ui/react` (this preset's choice, not Radix)
- Theme tokens (colors, radius, sidebar/chart colors) added to `app/globals.css`
  under `:root`, `.dark`, and `@theme inline`
- `app/layout.tsx` updated to load the `DM Sans` font and apply `font-sans`

### Notes
- Original custom tokens (`--bg`, `--fg`, `--muted`, `--accent`) still exist
  alongside the new shadcn tokens (`--background`, `--foreground`, etc.) in
  `globals.css` — worth reconciling so there's one source of truth for colors.
- Add new components with `npx shadcn@latest add <component>`; log them here as
  they're added.
- Record any theme token changes (brand colors, typography, radius) here.
